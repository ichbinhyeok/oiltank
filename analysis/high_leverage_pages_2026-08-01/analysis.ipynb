- Build `/heating-oil-usage-calculator/` and `/basement-oil-tank-removal/` next.
- Promote the existing removal-cost and leak guides instead of creating duplicates.

## Context & Methods

Decision: which additional pages create the most traffic leverage without thin-page expansion or intent cannibalization.

### Key Assumptions

- Google Ads demand bands are directional and close variants overlap.
- SERP opportunity is a reviewed 2026-08-01 snapshot, not a rank guarantee.
- Weighted score = demand 30% + SERP opportunity 25% + current-data reuse 20% + distinct job 15% + commercial alignment 10%.
- Traffic acquisition is weighted more heavily than lead intent.

## Data

The candidate file is a bounded decision table built from the repository keyword exports, current route inventory, existing guide indexability, and current SERP spot checks.

```python
from pathlib import Path
import pandas as pd

source = Path("page_candidates.csv")
if not source.exists():
    source = Path("analysis/high_leverage_pages_2026-08-01/page_candidates.csv")
candidates = pd.read_csv(source)
candidates
```

## Results

```python
weights = {
    "demand_score": 0.30,
    "serp_opportunity": 0.25,
    "reuse_score": 0.20,
    "intent_distinctness": 0.15,
    "commercial_alignment": 0.10,
}
candidates["weighted_score"] = sum(candidates[column] * weight for column, weight in weights.items())
ranked = candidates.sort_values(["weighted_score", "candidate"], ascending=[False, True]).reset_index(drop=True)
ranked.insert(0, "rank", ranked.index + 1)
ranked[["rank", "candidate", "route_action", "weighted_score", "recommendation"]]
```

```python
assert candidates["candidate"].is_unique
assert candidates[list(weights)].notna().all().all()
assert candidates[list(weights)].apply(lambda column: column.between(1, 5).all()).all()
assert abs(sum(weights.values()) - 1.0) < 1e-9
assert ranked.iloc[0]["candidate"] == "Heating oil usage calculator"
assert ranked.iloc[1]["candidate"] == "Basement oil tank removal"
print("Validation passed: unique candidates, complete 1-5 inputs, weights sum to 1, expected top two preserved.")
```

```python
current_product_pages = 7  # home + six core routes
phase_one_new_urls = 2
phase_one_promoted_guides = 2
phase_one_surface = current_product_pages + phase_one_new_urls + phase_one_promoted_guides
phase_two_surface = phase_one_surface + 1  # replacement-cost
{"current": current_product_pages, "phase_one": phase_one_surface, "with_replacement_cost": phase_two_surface}
```

## Takeaways

1. Usage is the strongest traffic-first extension because it reuses verified gauge gallons and creates a separate recurring job: days remaining.
2. Basement removal is the best new commercial page because its intent is distinct and its SERP is less consolidated than replacement cost.
3. Removal cost and leak already have repository URLs. The leverage comes from rebuilding and indexing them, not duplicating them.
4. Replacement cost has the largest demand band but the hardest SERP and the highest evidence burden.
5. Lifespan, repair, and gauge troubleshooting should first expand current pages; split them only when Search Console shows a separate query cluster.
